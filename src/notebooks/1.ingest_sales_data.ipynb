```python
dbutils.widgets.text("catalog_name","")
catalog_name = dbutils.widgets.get("catalog_name")
```

```python
input_path = f"/Volumes/{catalog_name}/landing/raw_sales_data/"
bronze_table = f"{catalog_name}.bronze.sales"
```

```python
df_sales = (
    spark.read
         .option("header", "true")
         .option("inferSchema", "true")
         .csv(input_path)
)
```

```python
from pyspark.sql.functions import current_timestamp
df_bronze = df_sales.withColumn(
    "load_timestamp",
    current_timestamp()
)
```

```python
df_bronze.write.mode("append").saveAsTable(bronze_table)
```
