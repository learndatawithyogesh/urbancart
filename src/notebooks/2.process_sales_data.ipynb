```python
dbutils.widgets.text("catalog_name","")
catalog_name = dbutils.widgets.get("catalog_name")
```

```python
bronze_table = f"{catalog_name}.bronze.sales"
silver_table = f"{catalog_name}.silver.sales"
```

```python
df_bronze = spark.table(bronze_table)
```

```python
df_silver = df_bronze.withColumnRenamed("product_name","product")
```

```python
from pyspark.sql.functions import col
df_silver = df_silver.filter(col("transaction_id").isNotNull() &col("store_id").isNotNull())
```

```python
df_silver.write.mode("overwrite").saveAsTable(silver_table)
```
