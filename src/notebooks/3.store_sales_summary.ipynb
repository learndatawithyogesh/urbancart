```python
dbutils.widgets.text("catalog_name","")
catalog_name = dbutils.widgets.get("catalog_name")
```

```python
silver_table = f"{catalog_name}.silver.sales"
gold_table = f"{catalog_name}.gold.store_sales_summary"
```

```python
df_sales = spark.table(silver_table)
```

```python
from pyspark.sql.functions import col,sum,countDistinct

df_store_summary = df_sales.withColumn("sales_amount",col("quantity") * col("unit_price"))\
    .groupBy("sales_date", "store_id")\
    .agg(
        countDistinct("transaction_id").alias("total_transactions"),
        sum("quantity").alias("total_quantity"),
        sum("sales_amount").alias("total_sales")
    )
```

```python
df_store_summary.write.mode("overwrite").saveAsTable(gold_table)
```
